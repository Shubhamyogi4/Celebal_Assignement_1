## Step 5: Handle Missing Values

```python
df["discount"] = df["discount"].fillna(0)
```

```python
df
```

```python
df["seller_name"] = df["seller_name"].fillna("unknown")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df["what_customers_said"] = df["what_customers_said"].fillna("no reviews")
```

```python
df["seller_information"] = df["seller_information"].fillna("no info")
```

```python
df
```

```python
df.isnull().sum()
```

## Step 6: Remove Unnecessary Columns

```python
df=df.drop(columns=["videos","variations"])
```

```python
df
```

## Step 7: Select Specific Columns

## Step 8: Filter Dataset

## Step 9: Remove Duplicate Records

```python
df=df.drop_duplicates(subset=["title"])
```

```python
df
```

```python
df.columns
```

## Step 10: Create Quantity Column

```python
df["quantity"] = 2
```

```python
df
```

## Step 11: Create Derived Column

```python
df['final_price'] = df['final_price'].astype(str)

df['final_price'] = df['final_price'].str.replace('₹', '', regex=False)
df['final_price'] = df['final_price'].str.replace(',', '', regex=False)
df['final_price'] = df['final_price'].str.replace('"', '', regex=False)
df['final_price'] = df['final_price'].str.extract(r'(\d+\.?\d*)')[0]

df['final_price'] = pd.to_numeric(df['final_price'])

df['total_amount'] = df['final_price'] * df['quantity']
```

```python
df
```

## Step 12: Save Cleaned Dataset

```python
df.to_csv("Celebal_Assignment_1.csv",index=False)
```
